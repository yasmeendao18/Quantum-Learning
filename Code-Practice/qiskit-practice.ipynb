Define single qubit

```python
# Define single qubit
from qiskit import QuantumCircuit 

# 1 qubit
qc = QuantumCircuit(1) 

print(qc.draw())
```

Add classical bit

```python
# add classical bit
from qiskit import QuantumCircuit

# 1 qubit and 1 classical bit
qc = QuantumCircuit(1,1)

print(qc.draw())
```

Apply gates

```python
# X gate (NOT) gate flips |0> to |1> 
# measurement irreversible 
qc.x(0) # single qubit gate
qc.measure(0, 0)

print(qc.draw())
```

```python
# hadamard gate
# superposition state where equal chance of 0 or 1 

qc.h(0) # single qubit gate
qc.measure(0,0)

print(qc.draw())
```

Measuring qubit - collapsing superposition 

```python
# Bell state superpositon and entaglement
# after mesaurement always 00 or 11, measuring qubit 0 determines qubit 1


qc = QuantumCircuit(2,2)

# superposition quibit 0
qc.h(0)
# CNOT entangle qubit 1 with qubit 0
qc.cx(0, 1) # two qubit gate
# measure both
qc.measure([0,1], [0,1])

print(qc.draw())
```

Verifying results on simulator

```python
from qiskit_aer import Aer
from qiskit import QuantumCircuit, transpile
qc = QuantumCircuit(1, 1)
qc.h(0)
qc.measure(0, 0)
sim = Aer.get_backend('aer_simulator')
# rewriting/compiling circuit so backend can run it (required before sim run)
compiled = transpile(qc, sim)
# more shots closer to theoretical 50/50
job = sim.run(compiled, shots=1000)
result = job.result()
print(result.get_counts())
```
